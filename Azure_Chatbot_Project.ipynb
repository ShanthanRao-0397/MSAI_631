```python
import requests

endpoint = "https://ai-chatbot-service-123.cognitiveservices.azure.com/"
api_key = "API_KEY"

url = endpoint + "text/analytics/v3.1/sentiment"

headers = {
    "Ocp-Apim-Subscription-Key": api_key,
    "Content-Type": "application/json"
}

def analyze_sentiment(text):
    documents = {
        "documents": [
            {"id": "1", "language": "en", "text": text}
        ]
    }

    response = requests.post(url, headers=headers, json=documents)
    return response.json()

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye!")
        break

    result = analyze_sentiment(user_input)

    sentiment = result["documents"][0]["sentiment"]

    if sentiment == "positive":
        print("Bot: I'm glad to hear that!")
    elif sentiment == "negative":
        print("Bot: I'm sorry you're feeling this way.")
    else:
        print("Bot: I understand.")
```
