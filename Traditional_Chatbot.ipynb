```python
import nltk
from nltk.chat.util import Chat, reflections

pairs = [
    [
        r"hi|hello|hey",
        ["Hello!", "Hi there!", "Hey!"]
    ],
    [
        r"what is your name ?",
        ["I am a traditional chatbot created using Python."]
    ],
    [
        r"how are you ?",
        ["I am doing well."]
    ],
    [
        r"sorry (.*)",
        ["No problem."]
    ],
    [
        r"quit",
        ["Bye! Take care."]
    ]
]
chatbot = Chat(pairs, reflections)

print("Hello! Type 'quit' to exit.")

chatbot.converse()
```
